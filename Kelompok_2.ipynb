```python
import cv2
import os
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
from google.colab import drive

drive.mount('/content/drive')

nama_file_dicari = "20242930011-6ZJKj.jpg"
root_drive = "/content/drive/MyDrive"

path_ditemukan = None
for root, dirs, files in os.walk(root_drive):
    if nama_file_dicari in files:
        path_ditemukan = os.path.join(root, nama_file_dicari)
        break

img = cv2.imread(path_ditemukan)
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8,8))
plt.imshow(imgRGB)
plt.grid(True)
plt.show()

imgLingkaran = imgRGB.copy()

cv2.circle(imgLingkaran, center=(140, 120), radius=20, color=(255,0,0), thickness=-1)
cv2.circle(imgLingkaran, center=(200, 120), radius=20, color=(0,255,0), thickness=-1)
cv2.ellipse(imgLingkaran, center=(170,180), axes=(30,18), angle=0, startAngle=0, endAngle=360, color=(0,0,255), thickness=-1)

plt.figure(figsize=(8,8))
plt.imshow(imgLingkaran)
plt.axis('off')
plt.show()

imgLingkaranBG = cv2.cvtColor(imgLingkaran, cv2.COLOR_RGB2BGR)
cv2.imwrite("/content/drive/MyDrive/Foto/Copy of 20242930011-6ZJKj.jpg", imgLingkaranBG)
```
